```python
import os

def get_sorted_md_files(directory):
    # Listar archivos .md
    files = [f for f in os.listdir(directory) if f.endswith('.md')]
    
    # Función para extraer el número del inicio del archivo para ordenar
    def get_number(filename):
        parts = filename.split()
        if parts and parts[0].isdigit():
            return int(parts[0])
        return float('inf') # Si no tiene número, va al final
    
    return sorted(files, key=get_number)

full_text = ""

# 1. Procesar archivos de la carpeta raíz (Plan Final)
current_dir = '.'
print(f"Leyendo archivos de: {os.path.abspath(current_dir)}")
files_root = get_sorted_md_files(current_dir)

for filename in files_root:
    print(f" - Agregando: {filename}")
    path = os.path.join(current_dir, filename)
    with open(path, 'r', encoding='utf-8') as f:
        full_text += f.read() + "\n\n"

# 2. Procesar archivos de la carpeta Interfaces
interfaces_dir = os.path.join(current_dir, 'Interfaces')
if os.path.exists(interfaces_dir):
    print(f"Leyendo archivos de: {os.path.abspath(interfaces_dir)}")
    files_interfaces = get_sorted_md_files(interfaces_dir)
    
    for filename in files_interfaces:
        print(f" - Agregando: {filename}")
        path = os.path.join(interfaces_dir, filename)
        with open(path, 'r', encoding='utf-8') as f:
            full_text += f.read() + "\n\n"

# Guardar el resultado
output_filename = 'Plan_Completo_Compilado.md'
with open(output_filename, 'w', encoding='utf-8') as f:
    f.write(full_text)

print(f"\n¡Listo! Todo el contenido se ha compilado en '{output_filename}'")
# print(full_text[:500]) # Descomentar para ver una vista previa
```
